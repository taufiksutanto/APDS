## 1. Why programs need conditional logic

A **conditional statement** selects an action according to a condition. Its
basic form is: **if a condition holds, perform an action**.

Explicit rules can form part of a rule-based system. An `if` statement by itself
does not learn from data and should not be described as machine learning.
It is a general programming construct used in many kinds of software.

In Python, a condition is evaluated for its truth value. A comparison such as
`x < 0` produces `True` or `False`; other objects can also be tested according to
Python's truth-value rules, which we will examine shortly.

### Special cases in mathematics and everyday reasoning

Algorithms frequently need different rules for different cases. For example,
2 is the only even prime. A mathematical expression may also need special
treatment where a denominator vanishes:

$$
f(n)=\begin{cases}
\dfrac{n^2-1}{n-1}, & n\ne 1,\\
2, & n=1.
\end{cases}
$$

The second branch defines a value at a point where the quotient is undefined.
A familiar everyday analogy is an activity permitted only under specified
circumstances; the relevant conditions must be stated explicitly.

For a more personal example, imagine planning a Sunday morning:

- **Heavy rain:** take a warm shower, prepare noodles and tea, and reflect on old memories.
- **Light rain:** have some bread and work on an assignment.
- **Otherwise:** enjoy a little more sleep.

Can you suggest another situation in mathematics, programming, or daily life
that requires different actions for different cases?

### From a verbal rule to a decision table

This self-contained table replaces a diagram referenced in the source notebook.

| Condition considered | If true | If false |
|---|---|---|
| Is the rain heavy? | Choose the heavy-rain plan. | Consider the next condition. |
| Is the rain light? | Choose the light-rain plan. | Choose the default plan. |

A decision table makes both the exceptional cases and the default action visible.
A flowchart can emphasize the paths; pseudocode can emphasize execution order.

## 2. Branch structure, indentation, and order

Python uses `elif`, rather than `else if`:

```python
if first_condition:
    first_action()
elif second_condition:
    second_action()
else:
    default_action()
```

The names above are placeholders, not functions defined in this notebook.
Each branch may contain more than one statement.

### Components of a conditional statement

| Component | Example | Role |
|---|---|---|
| Keyword | `if`, `elif`, `else` | Introduces a branch. |
| Condition | `temperature < 20` | Determines whether a conditional branch is selected. |
| Colon | `:` | Ends the branch header. |
| Indented body | `print("Cool")` | Contains statements executed for that branch. |
| Unindented continuation | `print("Finished")` | Continues after the whole decision. |

Use consistent indentation, conventionally four spaces for each level. Related
`if`, `elif`, and `else` headers align vertically. The indentation also determines
which `if` owns an `else` in a nested structure.

### How Python selects a branch

- Conditions in a single chain are considered from top to bottom.
- The first truthy condition selects its body; later conditions in that chain
  are not evaluated.
- `elif` and `else` are optional, and `else` has no condition.
- Without `else`, no body may run when all conditions are false.
- With `else`, one body is selected if evaluation reaches it normally.

These rules concern **one chain**. A separate `if` statement starts another
decision, so multiple independent bodies can run.

````python
temperature = 18
if temperature < 20:
    print("The temperature is below 20 degrees.")
    print("This second statement belongs to the same branch.")
print("The temperature check is complete.")

number = 0
if number < 0:
    category = "negative"
elif number == 0:
    category = "zero"
else:
    category = "positive"
print("Number category:", category)
````

### Branch ordering: broad rules can hide specific cases

The source referred to a missing image illustrating a hierarchy error. The
following example provides an explicit replacement. These thresholds are
fictional teaching rules, not an institutional grading policy.

If every score of at least 60 is accepted by the first branch, a later branch
for scores of at least 80 is unreachable. Put the more specific threshold first.

````python
score = 85

# Deliberately incorrect ordering: the second branch cannot be selected.
if score >= 60:
    incorrect_label = "Pass"
elif score >= 80:
    incorrect_label = "High performance"
else:
    incorrect_label = "Below threshold"

# Correct ordering: consider the higher threshold first.
if score >= 80:
    correct_label = "High performance"
elif score >= 60:
    correct_label = "Pass"
else:
    correct_label = "Below threshold"

print("Incorrect ordering:", incorrect_label)
print("Correct ordering:", correct_label)
````

single selected category.

````python
number = 6
if number > 0:
    print("The number is positive.")
if number % 2 == 0:
    print("The number is even.")
````

## 3. Comparisons and truth values

A condition need not already have type `bool`. Python applies truth-value
testing, so `if items:` is a conventional way to ask whether a built-in list is
nonempty. This is intentional language behavior, not an exception to the syntax.

Nevertheless, choose a condition that expresses your intended question. For
example, `if score:` rejects a score of zero as false, even when zero is valid.
Use `score is not None` when you specifically mean “a value was supplied.”

### Comparison, membership, and identity

| Operator | Meaning | Example |
|---|---|---|
| `==`, `!=` | Equal to; not equal to | `x == 2`, `x != 2` |
| `<`, `<=`, `>`, `>=` | Ordered comparisons | `score >= 60` |
| `in`, `not in` | Membership; nonmembership | `status in {"ready", "done"}` |
| `is`, `is not` | Object identity | `value is None` |

`=` assigns a value; `==` compares values. Use `is None` for the special missing
sentinel, and use `==` to compare numbers or text. Python also permits chained
comparisons such as `0 <= score <= 100`.

````python
score = 75
status = "ready"
missing_value = None
print("Within the allowed interval:", 0 <= score <= 100)
print("Recognized status:", status in {"ready", "done"})
print("Value is missing:", missing_value is None)

first_list = [1, 2]
second_list = [1, 2]
alias = first_list
print("Equal contents:", first_list == second_list)
print("Same object:", first_list is second_list)
print("Alias is the same object:", first_list is alias)
````

### Python's truth-value rules

For the built-in objects relevant here, falsy examples include:

- `False` and `None`;
- numeric zeros such as `0`, `0.0`, and `0j`;
- empty strings and containers such as `""`, `[]`, `()`, `{}`, and `set()`.

Nonzero numbers and nonempty strings or containers are truthy. In particular,
the strings `"False"` and `"0"` are truthy because they are not empty.
These rules should not be generalized blindly to every third-party object:
custom objects can define their own truth behavior.

See the official [truth-value documentation](https://docs.python.org/3/library/stdtypes.html#truth-value-testing).

````python
value = 1
if value:
    print("value is interpreted as true.")
else:
    print("value is interpreted as false.")
````

````python
examples = [False, None, 0, 0.0, 0j, "", [], (), {}, set(), -2, "False", "0", [0]]
print(f"{'Object':<10} {'Truth value'}")
for value in examples:
    print(f"{repr(value):<10} {bool(value)}")

score = 0
print("Truthiness of score:", bool(score))
print("Score was supplied:", score is not None)
````

## 4. Combining conditions

Use `and`, `or`, and `not` for logical operations. `not` is Python's logical
negation keyword; `!` by itself is not a Python negation operator. The symbol
`!=` is a single operator meaning “not equal.”

| A | B | A and B | A or B | not A | Boolean XOR |
|---|---|---|---|---|---|
| False | False | False | False | True | False |
| False | True | False | True | True | True |
| True | False | False | True | False | True |
| True | True | True | True | False | False |

This table concerns Boolean operands. In general, `and` and `or` return one of
their operands, rather than always returning an object of type `bool`.

````python
a = True
b = False
print("Logical AND:", a and b)
print("Logical OR:", a or b)
print("Logical NOT:", not a)

# On Boolean operands these results match, but the operators have different roles.
print("Boolean &: ", a & b)
print("Boolean |: ", a | b)
````

### Logical operators are not interchangeable with bitwise operators

For integers, `&`, `|`, and `^` operate on binary representations. They do not
short-circuit. Ordinary scalar conditions are usually clearer with `and`, `or`,
and `not`. The examples below expose the difference rather than treating the
two operator families as equivalent.

For Boolean inputs, XOR is true when exactly one operand is true. You can write
`a != b` or `a ^ b` when both operands are Boolean.

````python
# Retained integer XOR example: 1 XOR 0 is 1, which converts to True.
a = 1
b = 0
print("XOR converted to bool:", bool(a ^ b))

left = True
right = False
print("Boolean XOR:", left != right)
print("Integer bitwise AND:", 2 & 1)
print("Logical AND returns an operand:", 2 and 1)
print("Fallback text:", "" or "anonymous")
````

### Short-circuit evaluation

`and` does not evaluate its right operand when the left operand is falsy.
`or` does not evaluate its right operand when the left operand is truthy.
Place a protective condition before an expression that requires it.

````python
numerator = 10
denominator = 0
safe_and_large = denominator != 0 and numerator / denominator > 2
print("A valid ratio exceeds 2:", safe_and_large)

optional_score = None
valid_score = optional_score is not None and 0 <= optional_score <= 100
print("A valid score was supplied:", valid_score)
````

### Precedence and two common traps

Among the operators used here, arithmetic is evaluated before comparisons,
followed by `not`, then `and`, then `or`. Parentheses make an intended grouping
explicit. De Morgan's laws give useful equivalent conditions:

- `not (a and b)` is equivalent in truth value to `(not a) or (not b)`.
- `not (a or b)` is equivalent in truth value to `(not a) and (not b)`.

Write both comparisons in `x == 3 or x == 5`, or use `x in (3, 5)`.
The expression `x == 3 or 5` contains a standalone truthy integer. Similarly,
`value or default` replaces **every falsy value**, including valid zeros; use an
explicit `None` check when zero must be preserved.

````python
x = 2
print("Incorrect condition interpreted as bool:", bool(x == 3 or 5))
print("Correct membership condition:", x in (3, 5))

supplied_count = 0
print("Truthiness-based fallback:", supplied_count or 10)
if supplied_count is None:
    chosen_count = 10
else:
    chosen_count = supplied_count
print("None-specific fallback:", chosen_count)

print("Without parentheses:", True or False and False)
print("With explicit grouping:", (True or False) and False)
````

### Guarding a mathematical expression

The opening example assigns $f(1)=2$ explicitly. Otherwise, it evaluates
$(n^2-1)/(n-1)$. The test must occur **before** division. Computing the quotient
first and checking the denominator afterward would be too late.

````python
for n in [0, 1, 2]:
    if n != 1:
        function_value = (n**2 - 1) / (n - 1)
    else:
        function_value = 2
    print(f"f({n}) = {function_value}")
````

The source also used a related example with a different special point:

$$
g(x)=\begin{cases}
\dfrac{x^2-4}{x-2}, & x\ne 2,\\
x+2, & x=2.
\end{cases}
$$

Notice that the denominator is zero at **2**, not at 1. In exact algebra,
$x^2-4=(x-2)(x+2)$, so the extended function equals $x+2$ everywhere. The
branching version is useful for instruction; the simplified expression is
clearer for this particular function and avoids cancellation near the special point.

````python
raw_x = "2"
# raw_x = input("Enter a real number x: ")
x = float(raw_x)
if x != 2:  # != means "not equal"; logical negation is written with not.
    function_value = (x**2 - 4) / (x - 2)
else:
    function_value = x + 2
print("Piecewise result:", function_value)
print("Simplified result:", x + 2)
````

**A note on numeric comparisons.** Equality with an explicitly supplied special
value differs from asking whether two computed floating-point results are
approximately equal. Binary floating-point arithmetic does not represent every
decimal exactly. For an approximate comparison, choose a tolerance appropriate
to the problem rather than changing every `==` test automatically.

````python
from math import isclose

result = 0.1 + 0.2
print("Computed value:", result)
print("Exact equality:", result == 0.3)
print("Approximate equality:", isclose(result, 0.3, rel_tol=1e-9, abs_tol=1e-12))
````

## 5. Nested decisions and conditional expressions

A nested `if` is a decision inside another decision. It is useful when a second
question makes sense only after a prerequisite is satisfied. In this example,
we distinguish a missing score from a supplied score outside the valid interval.
Each `else` aligns with the `if` to which it belongs.

````python
for score in [75, None, 150]:
    if score is not None:
        if 0 <= score <= 100:
            message = "Valid score"
        else:
            message = "Score outside the allowed interval"
    else:
        message = "Missing score"
    print(f"{score!r}: {message}")
````

When several nested branches simply assign a value, an `elif` chain can often
reduce indentation. Retain nesting when it makes the dependency clearer; do not
flatten a decision merely to save lines.

For a short choice between two expressions, Python also permits
`value_if_true if condition else value_if_false`. Only the selected expression
is evaluated. Prefer a full statement for multiple actions or complex logic.

````python
number = 7
parity = "even" if number % 2 == 0 else "odd"
print(f"{number} is {parity}.")

score = None
if score is None:
    message = "Missing score"
elif 0 <= score <= 100:
    message = "Valid score"
else:
    message = "Score outside the allowed interval"
print(message)
````

## 6. Combining `for` and `if`

A loop visits candidates; a condition decides what to do with each candidate.
Let us select integers from 1 through 24 that are divisible by 3 **or** 5.
The remainder operator `%` lets us test divisibility.

````python
selected_values = []
for number in range(1, 25):
    if number % 3 == 0 or number % 5 == 0:
        selected_values.append(number)
print(selected_values)
````

The following filtering comprehension has the same behavior. A value such as
15 is appended only once: one candidate is visited once, even though both parts
of the condition are true.

Compare a **filter**, `[x for x in values if condition]`, with a **conditional
transformation**, `[a if condition else b for x in values]`. A filter can reduce
the number of items; a transformation supplies one result for each input item.

````python
selected_values = [
    number for number in range(1, 25)
    if number % 3 == 0 or number % 5 == 0
]
print(selected_values)

labels = ["even" if number % 2 == 0 else "odd" for number in range(1, 6)]
print(labels)
````

### The prime-number example: a direct implementation

A prime is an integer greater than 1 whose only positive divisors are 1 and
itself. Values below 2 are not prime; 2 is prime. For larger candidates, we can
search for a divisor and use `break` once we find one.

The loop below tests 2 through `n - 1`, because the stop value of `range()` is
excluded. `break` exits the nearest enclosing loop, not the surrounding `if`.

````python
raw_n = "29"
# raw_n = input("Enter an integer n: ")
n = int(raw_n)

if n < 2:
    prime = False
elif n == 2:
    prime = True
else:
    prime = True
    for divisor in range(2, n):
        if n % divisor == 0:
            prime = False
            break
print(f"{n} is prime: {prime}")
````

### A more efficient bound and important test cases

If a positive composite integer $n=ab$ had both factors greater than $\sqrt{n}$,
their product would exceed $n$. We therefore only need to search through
$\lfloor\sqrt{n}\rfloor$. `math.isqrt()` computes this integer bound without
converting a large integer to a floating-point number.

Add 1 to include the bound in `range()`. Perfect squares such as 25 and 49 are
useful tests: omitting that endpoint would miss their square-root divisors.

````python
from math import isqrt

prime_results = {}
for candidate in [-3, 0, 1, 2, 3, 4, 9, 25, 29, 49]:
    if candidate < 2:
        prime = False
    else:
        prime = True
        for divisor in range(2, isqrt(candidate) + 1):
            if candidate % divisor == 0:
                prime = False
                break
    prime_results[candidate] = prime
    print(f"{candidate:3d}: {prime}")
````

## 7. Validation and exception handling

Programs should distinguish input that violates an intended rule from an
operation that fails. A **domain constraint** specifies admissible values;
exception handling provides a controlled response to a raised error.

- Use `if` for an explicit rule, such as `1 <= n <= 10`.
- Use `try`–`except` around an operation that may fail, such as `int(text)`.
- Catch the specific errors you expect, and keep the protected block focused.
- Continue with a parsed variable only after conversion succeeds.

A prompt saying “between 1 and 10” does not enforce that interval. A bare
`except:` can also hide unrelated failures and intercept interruption signals.
Prefer explicit exception classes. See the official
[errors and exceptions tutorial](https://docs.python.org/3/tutorial/errors.html).

### A corrected version of the source example

We require an integer in $[1,10]$ and evaluate

$$h(n)=\frac{n^2-1}{(n-1)(n-5)(n-7)}.$$

The original expression is undefined at 1, 5, and 7. Here we retain that domain;
we do not silently assign a value at 1 after algebraic cancellation.

`except ValueError` handles conversion errors. The `else` attached to `try`
runs only when the protected conversion succeeds, so a previous or undefined
value of `n` cannot be used accidentally.

````python
raw_n = "5"
# raw_n = input("Enter an integer from 1 to 10: ")
try:
    n = int(raw_n)
except ValueError:
    print("The input is not a valid integer.")
else:
    if not 1 <= n <= 10:
        print("The integer must be between 1 and 10.")
    elif n in (1, 5, 7):
        print("The function is undefined because the denominator is zero.")
    else:
        function_value = (n**2 - 1) / ((n - 1) * (n - 5) * (n - 7))
        print("Function value:", function_value)
````

### Catch an operational error explicitly

When division itself is the operation under discussion, catch
`ZeroDivisionError`. In the preceding mathematical example, checking known
excluded inputs first expresses the domain particularly clearly.

A `finally` block runs when control leaves the `try` statement during normal
Python execution, whether or not an exception was handled. It is commonly used
for cleanup; it is not a replacement for `else` or input validation.

````python
for denominator in [2, 0]:
    try:
        quotient = 10 / denominator
    except ZeroDivisionError:
        print("Cannot divide by zero.")
    else:
        print("Quotient:", quotient)
    finally:
        print("Division attempt complete.")
````

### Check conversion, domain, and computation separately

The following cases cover invalid text, an out-of-range integer, excluded
points, and ordinary valid values. `"3.5"` is not accepted by `int()`; converting
through `float()` and then truncating would change the meaning of “integer input.”

````python
validation_results = []
for text in ["hello", "3.5", "0", "1", "5", "7", "2", "10"]:
    try:
        n = int(text)
    except ValueError:
        outcome = "not an integer"
    else:
        if not 1 <= n <= 10:
            outcome = "outside [1, 10]"
        elif n in (1, 5, 7):
            outcome = "undefined denominator"
        else:
            value = (n**2 - 1) / ((n - 1) * (n - 5) * (n - 7))
            outcome = f"valid result: {value:.6f}"
    validation_results.append(outcome)
    print(f"{text!r:>8} -> {outcome}")
````

## 8. A complete median example

The source develops a median calculation through several input-processing
cells. We will retain that progression and make its assumptions explicit.
For now, the input is a nonempty sequence of integers. Text conversion and
empty-input handling will be combined in a reusable version afterward.

**Step 1:** split a space-separated string into tokens, convert each token with
`int()`, then sort the resulting list. `.sort()` changes the list in place and
returns `None`; do not assign its return value back to the list.

````python
raw_values = "3 1 2 7"
# raw_values = input("Enter integers separated by spaces: ")
tokens = raw_values.split()
values = [int(token) for token in tokens]
values.sort()
print("Sorted values:", values)
````

````python
size = len(values)
print("Number of observations:", size)
````

**Step 2:** repeat the preparation with the second sample used in the source.
`split()` handles repeated spaces and other whitespace, not only a single space.

````python
raw_values = "23 4 5 1"
# raw_values = input("Enter integers separated by spaces: ")
values = [int(token) for token in raw_values.split()]
values.sort()
size = len(values)
print("Sorted values:", values)
print("Number of observations:", size)
````

**Step 3:** let `middle = size // 2`, using integer division.

| Case | Zero-based position(s) | Median |
|---|---|---|
| Odd length | `middle` | The single middle value. |
| Even length | `middle - 1`, `middle` | The mean of the two central values. |
| Empty list | No valid middle position | Undefined; handle explicitly. |

For `[1, 4, 5, 23]`, the central values are 4 and 5, so the median is **4.5**.

````python
if size == 0:
    median_value = None
    print("The median is undefined for an empty list.")
elif size % 2 == 0:
    middle = size // 2
    median_value = (values[middle - 1] + values[middle]) / 2
    print("Median:", median_value)
else:
    middle = size // 2
    median_value = values[middle]
    print("Median:", median_value)
````

### Make the calculation reusable

A function groups reusable instructions. `def` introduces it, its parameter
receives the input, and `return` sends the result back to the caller. Here,
`raise ValueError(...)` deliberately reports a violated input requirement.

This function assumes finite numeric values that can be ordered and averaged.
It makes a sorted copy, preserving the caller's collection, and rejects empty
input. Missing values and non-finite numbers require an explicit data policy;
they should not be accepted silently as ordinary observations.

````python
def median_of_values(numbers):
    """Return the median of nonempty, finite numeric observations."""
    ordered = sorted(numbers)
    if not ordered:
        raise ValueError("The median requires at least one observation.")
    middle = len(ordered) // 2
    if len(ordered) % 2 == 0:
        return (ordered[middle - 1] + ordered[middle]) / 2
    return ordered[middle]

original_values = [23, 4, 5, 1]
print("Median:", median_of_values(original_values))
print("Input preserved:", original_values)
````

### Combine parsing and median calculation

This small interface accepts **integers** separated by whitespace. Blank input
is rejected explicitly. An invalid token raises `ValueError` during conversion;
a numeric input with one value has that value as its median.

````python
def median_from_text(text):
    """Parse whitespace-separated integers and return their median."""
    tokens = text.split()
    if not tokens:
        raise ValueError("Enter at least one integer.")
    numbers = [int(token) for token in tokens]
    return median_of_values(numbers)

for text in ["3 1 2 7", "9 1 4", "8", "", "2 cat 4"]:
    try:
        result = median_from_text(text)
    except ValueError as error:
        print(f"{text!r}: invalid input ({error})")
    else:
        print(f"{text!r}: median = {result}")
````

### Solution 1 · Identify the unreachable branch

The original algorithm produces `[1]`, `[2, 2, 2]`, and `[5, 5, 5]` for inputs
1, 2, and 5, respectively. The special branch for 5 is never reached because
`5 >= 2` is already true. Testing `n == 5` first makes that case produce `[7]`.
For 0, `range(1, 1)` is empty, so the original algorithm prints nothing.

````python
original_traces = {}
corrected_traces = {}
for n in [0, 1, 2, 5]:
    original_output = []
    if n >= 2:
        for _ in range(1, 4):
            original_output.append(n)
    elif n == 5:  # Deliberately retained unreachable branch for comparison.
        original_output.append(n + 2)
    else:
        for i in range(1, n + 1):
            original_output.append(i)

    corrected_output = []
    if n == 5:
        corrected_output.append(n + 2)
    elif n >= 2:
        for _ in range(1, 4):
            corrected_output.append(n)
    else:
        for i in range(1, n + 1):
            corrected_output.append(i)

    original_traces[n] = original_output
    corrected_traces[n] = corrected_output
    print(f"n={n}: original={original_output}, corrected={corrected_output}")
````

### Solution 2 · Independent conditions report all properties

For 6, the chain reports only `positive`; separate conditions report both
`positive` and `even`. For -2 and 0, both designs report only `even`.

````python
for x in [6, -2, 0]:
    chain_labels = []
    if x > 0:
        chain_labels.append("positive")
    elif x % 2 == 0:
        chain_labels.append("even")

    independent_labels = []
    if x > 0:
        independent_labels.append("positive")
    if x % 2 == 0:
        independent_labels.append("even")
    print(f"x={x}: chain={chain_labels}, independent={independent_labels}")
````

### Solution 3 · Check the reusable median implementation

The `median_of_values()` and `median_from_text()` functions in Section 8 are

`assert` raises an error if its condition is false. Use checks to examine

````python
from statistics import median as standard_median

median_cases = [
    ([3, 1, 2, 7], 2.5),
    ([9, 1, 4], 4),
    ([8], 8),
    ([-5, -1, -3], -3),
    ([2, 2, 2, 2], 2),
]
for numbers, expected in median_cases:
    before = numbers.copy()
    result = median_of_values(numbers)
    assert result == expected == standard_median(numbers)
    assert numbers == before
    print(f"{numbers} -> {result}")

try:
    median_of_values([])
except ValueError as error:
    print("Empty input handled:", error)
else:
    raise AssertionError("Empty input should have been rejected.")

for text in ["", "2 cat 4", "1.5 2"]:
    try:
        median_from_text(text)
    except ValueError:
        print(f"Malformed or empty input rejected: {text!r}")
    else:
        raise AssertionError("This input should have been rejected.")
````

### Solution 5 · Validate before applying a threshold

The expected labels are `negative`, `negative`, `positive`, `positive`,
`positive`, `missing`, `invalid`, and `invalid`. The threshold includes 0.5.
The model probabilities here are synthetic, and the threshold is a teaching
choice rather than a recommended operating threshold for a real model.

````python
def label_probability(value):
    """Classify a finite numeric probability, or report missing/invalid input."""
    if value is None:
        return "missing"
    elif not 0 <= value <= 1:
        return "invalid"
    elif value >= 0.5:
        return "positive"
    else:
        return "negative"

solution_probabilities = [0.0, 0.49, 0.5, 0.85, 1.0, None, -0.1, 1.1]
probability_labels = []
for probability in solution_probabilities:
    label = label_probability(probability)
    probability_labels.append(label)
    print(f"{probability!r:>5}: {label}")
````

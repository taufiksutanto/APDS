# <center><font color="blue"> A Brief History of Python and Its Applications

<p>Python was introduced around 1980 by <strong>Guido van Rossum</strong>.</p>

<p>Python has a system that is, in some respects, &quot;similar&quot; to Java: it is multiplatform and has a kernel. # <center><font color="blue"> A Brief Introduction to Jupyter Notebook

* In this context, a "Notebook" is an interactive IDE that combines an editor, output display, and other useful functions.
* There are three main types of cells: Code, Markdown, and Raw.
    * Code: <font color="green">**Esc + Y**</font> &nbsp;==>&nbsp; We write Python programs in code cells.
    * Markdown: <font color="green">**Esc + M**</font> &nbsp;==>&nbsp; Similar to HTML, Markdown allows us to write formatted text much like a document in Microsoft Word.
    * Execute the current cell with <font color="green">**Ctrl + Enter**</font>.
    * Execute the current cell and create a new cell below it with <font color="green">**Alt + Enter**</font>.
* Markdown supports, among other things:
    - `#`, `##`, and `###` for Heading 1, Heading 2, and Heading 3
    - `**` and `*` for bold and italic text
    - `1, 2, 3` and `-` at the beginning of lines for numbered and bulleted lists
    - LaTeX commands for mathematical notation
* Jupyter Notebook can also be used with other programming languages, including Julia, Python, and R.

# <center><font color="blue"> Introduction to Modules in Python

* Modules are loaded using the **import** statement.
* We can also load selected functions from a module using the **from** statement.
* A loaded module can be assigned an alternative name using the **as** statement.
* Functions and modules that have been loaded are stored in Python's **namespace**.
* We can create a collection of functions in a `.py` file and then load that file in the same way as a regular module.
* In Python, modules are not necessarily written entirely in Python. For example, NumPy is written partly in C and Fortran, while libraries such as spaCy and Gensim use C/C++ components. This is one reason Python can achieve performance that is competitive with other programming languages.

# <center><font color="blue"> Basic Python Syntax

Consistent with its underlying design philosophy, Python is one of the programming languages that is relatively easy to learn, and its code can be written in a clear and well-structured manner. Several characteristics support this:

* Variables do not need to be declared before they are used.
* A semicolon (`;`) is not required at the end of each line.
* Curly braces `{}` are not required for several types of instructions.
* Python uses indentation (a tab or four spaces), which encourages code to be written in an orderly and readable structure.

# <center><font color="blue"> Basic Data Structures in Python

The basic (primitive) data structures in Python include:

* Atomic: Boolean, integer, float, string
* Non-atomic: Tuple, List, Set, Dictionary

**Notes:**
- Every variable in Python is a class/object with its own properties and functions.
- We can inspect these properties and functions using the `Tab` key or the built-in `dir(var)` command.

<center><img alt="" src="images/Python-data-structure.jpg" style="width: 600px;" />


```python
# Demonstration of basic syntax ~ A single-line comment begins with the hash symbol (#).
"""
Three double quotation marks can be used for a multi-line string,
as illustrated here. In instructional notebooks, this is often used
to provide longer explanatory notes.
"""
# Python, like MATLAB, does not require variable declarations and allows "variable juggling".
# In practice, however, this is strongly discouraged because it can reduce code clarity.
A = 3 # integer
B = 2.5 # float
C = True # Boolean: True or False
D = 'ABC soy sauce' # string; either single or double quotation marks may be used
N = None # a special atomic value in Python, similar to NULL in a database, but different from NaN

print("C = {}, A = {}".format(C, A)) 
# The print command displays output. By using the "format" syntax, we do not need to specify the type of variable C.
D # this line will not appear in the output
B, A # in Jupyter Notebook, the final expression in each cell is displayed automatically
```

```python
"""
Use the "type" command to determine the type of a variable.
Use "dir" to inspect its available functions and properties.
These two commands are particularly important in Data Science, ML, and AI. Why?
"""
print(type(D))
print(dir(D)) # What happens if we omit the print command on this line?
```
